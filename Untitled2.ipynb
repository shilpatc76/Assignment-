```python
import pandas as pd

retail1 = pd.read_excel("USECASE - Data Engineering.xlsx", sheet_name="retail_data1")
retail2 = pd.read_excel("USECASE - Data Engineering.xlsx", sheet_name="retail_data2")

df = pd.concat([retail1, retail2], ignore_index=True)

# Remove Duplicate Records

print("Rows before removing duplicates:", df.shape[0])

df = df.drop_duplicates()

print("Rows after removing duplicates:", df.shape[0])

# Check Null Values
print("\nNull Values Before Cleaning:")
print(df.isnull().sum())

# Handle Null Values
# Fill missing prices with median price

df['price'] = df['price'].fillna(df['price'].median())

# Fill missing quantity with 1
df['quantity'] = df['quantity'].fillna(1)

# Fill missing discount with 0
df['discount'] = df['discount'].fillna(0)

# Fill missing city with Unknown
df['city'] = df['city'].fillna('Unknown')

# Remove rows where important fields are missing
df = df.dropna(subset=[
    'transaction_id',
    'customer_id',
    'product_id',
    'transaction_date'
])

# Verify Null Values

print("\nNull Values After Cleaning:")
print(df.isnull().sum())

# Save cleaned file
df.to_csv("cleaned_retail_data.csv", index=False)

print("Data Cleaning Completed Successfully")
```

**Standardizing the names**

```python


df['product_name'] = (
    df['product_name']
    .astype(str)
    .str.strip()
    .str.title()
)

# ==================================
# Standardize Category Names
# ==================================

category_mapping = {
    'electronics': 'Electronics',
    'elec': 'Electronics',
    'home appliances': 'Home Appliances',
    'home': 'Home Appliances',
    'furniture': 'Furniture',
    'furn': 'Furniture',
    'clothing': 'Clothing',
    'cloth': 'Clothing'
}

df['category'] = (
    df['category']
    .astype(str)
    .str.lower()
    .map(category_mapping)
)

# ==================================
# Convert Date Format
# ==================================

df['transaction_date'] = pd.to_datetime(
    df['transaction_date'],
    errors='coerce'
)

# Remove invalid dates
df = df[df['transaction_date'].notna()]

# ==================================
# Remove Invalid Quantities
# ==================================

df = df[df['quantity'] > 0]

# ==================================
# Create Date Columns
# ==================================

df['year'] = df['transaction_date'].dt.year
df['month'] = df['transaction_date'].dt.month_name()
df['quarter'] = df['transaction_date'].dt.quarter

print("Data Standardization Completed")
```

**PII Masking**

```python
import hashlib

# Email Masking
def mask_email(email):
    if pd.isna(email):
        return None

    email = str(email)

    if '@' not in email:
        return email

    username, domain = email.split('@')
    return username[:2] + '****@' + domain

# Phone Hashing
def hash_phone(phone):
    if pd.isna(phone):
        return None

    return hashlib.sha256(
        str(phone).encode()
    ).hexdigest()

df['email'] = df['email'].apply(mask_email)
df['phone'] = df['phone'].apply(hash_phone)

print("PII Masking Completed")
```

**Merge with Product Details**

```python
product_df = pd.read_excel(
    "USECASE - Data Engineering.xlsx",
    sheet_name="product_details"
)

final_df = pd.merge(
    df,
    product_df,
    on='product_id',
    how='left',
    suffixes=('', '_master')
)

print("Product Table Joined Successfully")
```

**KPI Calculations**

```python
# Check data types first
print(final_df[['price', 'quantity', 'discount']].dtypes)

# Convert to numeric if required
final_df['price'] = pd.to_numeric(final_df['price'], errors='coerce')
final_df['quantity'] = pd.to_numeric(final_df['quantity'], errors='coerce')
final_df['discount'] = pd.to_numeric(final_df['discount'], errors='coerce')

# Fill missing discount values
final_df['discount'] = final_df['discount'].fillna(0)

# Create revenue column
final_df['revenue'] = (
    final_df['price'] *
    final_df['quantity'] *
    (1 - final_df['discount'])
)

# Verify
print(final_df[['price','quantity','discount','revenue']].head())
```

```python
total_revenue = final_df['revenue'].sum()

revenue_by_category = (
    final_df.groupby('category')['revenue']
    .sum()
    .reset_index()
)

revenue_by_city = (
    final_df.groupby('city')['revenue']
    .sum()
    .reset_index()
)

print("Total Revenue:", total_revenue)
print(revenue_by_category.head())
print(revenue_by_city.head())
```

```python
# Total Revenue
total_revenue = final_df['revenue'].sum()

# Total Orders
total_orders = final_df['transaction_id'].nunique()

# Average Order Value
avg_order_value = total_revenue / total_orders

# Revenue By Category
revenue_by_category = (
    final_df.groupby('category')['revenue']
    .sum()
    .reset_index()
    .sort_values('revenue', ascending=False)
)

# Revenue By City
revenue_by_city = (
    final_df.groupby('city')['revenue']
    .sum()
    .reset_index()
    .sort_values('revenue', ascending=False)
)

# Top Selling Products
top_products = (
    final_df.groupby('product_name')['revenue']
    .sum()
    .reset_index()
    .sort_values('revenue', ascending=False)
)

print("Total Revenue:", total_revenue)
print("Total Orders:", total_orders)
print("Average Order Value:", avg_order_value)
```

```python
final_df.to_csv("cleaned_retail_dataset.csv", index=False)

revenue_by_category.to_csv(
    "revenue_by_category.csv",
    index=False
)

revenue_by_city.to_csv(
    "revenue_by_city.csv",
    index=False
)

top_products.to_csv(
    "top_products.csv",
    index=False
)
```
